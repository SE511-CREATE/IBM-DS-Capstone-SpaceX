# SpaceX Data Wrangling
Goal: explore the collected data, understand the landing outcomes and create the target label `Class` (1 = first stage landed, 0 = did not).

```python
import os
import pandas as pd, numpy as np
```

```python
if os.path.exists('dataset_part_1.csv'):
    df = pd.read_csv('dataset_part_1.csv')
else:
    df = pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_1.csv')
df.head(10)
```

## 1. Missing values and data types

```python
print(df.isnull().sum() / len(df) * 100)
df.dtypes
```

## 2. Launches per site, orbit and landing outcome

```python
df['LaunchSite'].value_counts()
```

```python
df['Orbit'].value_counts()
```

```python
landing_outcomes = df['Outcome'].value_counts()
landing_outcomes
```

```python
for i, outcome in enumerate(landing_outcomes.keys()):
    print(i, outcome)
```

## 3. Create the landing-outcome label

```python
bad_outcomes = set(landing_outcomes.keys()[[1, 3, 5, 6, 7]])
bad_outcomes
```

```python
bad_outcomes = {'False ASDS', 'False Ocean', 'False RTLS', 'None ASDS', 'None None'}
df['Class'] = [0 if o in bad_outcomes else 1 for o in df['Outcome']]
df[['Outcome', 'Class']].head(10)
```

```python
print('Overall success rate:', df['Class'].mean())
df['Class'].value_counts()
```
